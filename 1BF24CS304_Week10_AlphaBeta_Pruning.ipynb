```python



def alpha_beta(node, depth, alpha, beta, is_max):
    if depth == 0 or isinstance(node, int):
        return node

    if is_max:
        best = float('-inf')

        for child in node:
            val = alpha_beta(child, depth-1,
                             alpha, beta, False)

            best = max(best, val)
            alpha = max(alpha, best)

            if alpha >= beta:
                print("Pruned:", node[node.index(child)+1:])
                break

        return best

    else:
        best = float('inf')

        for child in node:
            val = alpha_beta(child, depth-1,
                             alpha, beta, True)

            best = min(best, val)
            beta = min(beta, best)

            if beta <= alpha:
                print("Pruned:", node[node.index(child)+1:])
                break

        return best


# Main program
tree = [
    [3, 5, 6],
    [9, 1, 2],
    [0, -1, 4]
]

result = alpha_beta(tree, 2, float('-inf'),
                    float('inf'), True)

print("Optimal Root Value:", result)

```
