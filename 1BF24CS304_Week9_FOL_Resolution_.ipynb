```python
# FOL Resolution

# Step 1: Knowledge Base in CNF
KB = [
    {"~Food(Peanuts)", "Likes(John,Peanuts)"},
    {"~Eats(Anil,Peanuts)", "~NotKilled(Anil)",
     "Food(Peanuts)"},
    {"Eats(Anil,Peanuts)"},
    {"Alive(Anil)"},
    {"~Alive(Anil)", "NotKilled(Anil)"},
    {"~NotKilled(Anil)", "Alive(Anil)"},
    {"Food(Apple)"},
    {"Food(Vegetables)"},
    {"~Eats(Anil,Peanuts)", "Eats(Harry,Peanuts)"}
]

# Step 2: Negate the query
QUERY = "~Likes(John,Peanuts)"
KB.append({QUERY})


# Step 3: Resolution function
def resolve(c1, c2):
    result = []

    for l1 in c1:
        for l2 in c2:

            if l1 == "~" + l2 or l2 == "~" + l1:

                r = (c1 - {l1}) | (c2 - {l2})

                # Ignore tautological clauses
                if not any(
                    "~" + x in r for x in r
                    if not x.startswith("~")
                ):
                    result.append(frozenset(r))

    return result


# Step 4: Apply resolution
def resolution(KB):

    clauses = set(frozenset(c) for c in KB)

    while True:

        new = set()
        pairs = list(clauses)

        for i in range(len(pairs)):
            for j in range(i + 1, len(pairs)):

                res = resolve(pairs[i], pairs[j])

                for r in res:

                    # Empty clause means contradiction
                    if not r:
                        return True

                    if r not in clauses:
                        new.add(r)

        if not new:
            return False

        clauses.update(new)


# Step 5: Display result
print("Knowledge Base Created")
print("Negated Query Added")
print("Applying Resolution...")

if resolution(KB):
    print("Empty Clause Found")
    print("Query Proved!")
    print("John likes peanuts.")
else:
    print("Query Not Proved")
```
